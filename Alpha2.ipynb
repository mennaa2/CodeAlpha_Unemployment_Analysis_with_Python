```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1 = pd.read_csv("/content/Unemployment in India.csv")
df2 = pd.read_csv("/content/Unemployment_Rate_upto_11_2020.csv")
```

```python
print("Dataset 1:")
print(df1.head())
```

```python
print("\nDataset 2:")
print(df2.head())
```

```python
print(df2.isnull().sum())
df2 = df2.dropna()
```

```python
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()
```

```python
df1['Date'] = pd.to_datetime(df1['Date'])
df2['Date'] = pd.to_datetime(df2['Date'])
```

```python
df1 = df1.dropna()
df2 = df2.dropna()
```

```python
df = pd.concat([df1, df2], ignore_index=True)

print("\nCleaned Data:")
print(df.head())
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
```

```python
print("\nStatistics:")
print(df.describe())
```

```python
before_covid = df[df['Year'] < 2020]
during_covid = df[df['Year'] == 2020]

print("Before Covid Avg:",
      before_covid["Estimated Unemployment Rate (%)"].mean())

print("During Covid Avg:",
      during_covid["Estimated Unemployment Rate (%)"].mean())
```

```python
plt.bar(
    ['Before Covid', 'During Covid'],
    [
        before_covid["Estimated Unemployment Rate (%)"].mean(),
        during_covid["Estimated Unemployment Rate (%)"].mean()
    ]
)

plt.title("Impact of Covid-19 on Unemployment")
plt.show()
```

```python
sns.lineplot(x="Month", y="Estimated Unemployment Rate (%)", data=df)
plt.title("Seasonal Trend")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.lineplot(x="Date", y="Estimated Unemployment Rate (%)", data=df)
plt.title("Unemployment Rate Over Time")
plt.xticks(rotation=45)
plt.show()
```

```python
df['Month'] = df['Date'].dt.month

sns.boxplot(x="Month", y="Estimated Unemployment Rate (%)", data=df)
plt.title("Monthly Trend")
plt.show()
```

```python
top_regions = df.groupby("Region")["Estimated Unemployment Rate (%)"].mean().sort_values(ascending=False)

print("\nTop Regions by Unemployment:")
print(top_regions.head())
```

```python
df['Year'] = df['Date'].dt.year
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="Year", y="Estimated Unemployment Rate (%)", data=df)
plt.title("Unemployment Rate Before vs During Covid")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.barplot(x="Region", y="Estimated Unemployment Rate (%)", data=df)
plt.xticks(rotation=90)
plt.title("Unemployment Rate by Region")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
pivot = df2.pivot_table(
    values='Estimated Unemployment Rate (%)',
    index='Month',
    columns='Year'
)

plt.figure(figsize=(10,6))
sns.heatmap(pivot, cmap="YlGnBu")
plt.title("Unemployment Heatmap (Month vs Year)")
plt.show()
```

```python
avg_rate = df["Estimated Unemployment Rate (%)"].mean()
max_rate = df["Estimated Unemployment Rate (%)"].max()

print("\nInsights:")
print("Average Unemployment Rate:", avg_rate)
print("Maximum Unemployment Rate:", max_rate)
```

```python
covid_data = df[df['Year'] == 2020]

print("\nCovid Period Average:",
      covid_data["Estimated Unemployment Rate (%)"].mean())
```

- There are clear differences between states in unemployment levels.
- Some states are more affected economically than others.
